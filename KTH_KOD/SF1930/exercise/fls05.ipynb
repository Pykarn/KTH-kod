```python
from scipy import stats
import numpy as np
```

Hierarkisk modell:

$$
\begin{split}
Y \mid X_1 = x_1, X_2 = x_2 &\sim \mathrm{N}(x_1 x_2, 1) \\
X_1 \mid \Theta = \theta &\sim \mathrm{N}(\theta, 3) \\
X_2 \mid \Theta = \theta &\sim \mathrm{N}(\theta/2, 10) \\
\Theta &\sim \mathrm{Gamma}(5, 4)
\end{split}
$$

med $Y \perp \Theta \mid X_1, X_2$ samt $X_1 \perp X_2 \mid \Theta$.

```python
# Antal utfall vi kommer simulera.
n = 100000

# Förbered vektorer där vi kan samla utfallen. Det i:te utfallet utgörs av
#
#    (thetas[i], x1s[i], x2s[i], ys[i])

thetas = []
x1s = []
x2s = []
ys = []
```

```python
for _ in range(n):
    # Först genererar vi theta. Notera att vi i kursen använder intensitetsparametrisering
    # men att SciPy använder skalparametrisering, så vi måste ta scale=1/4.
    theta = stats.gamma.rvs(a=5, scale=1/4)
    
    # Givet theta kan vi sedan generera x1 och x2 (notera igen skala, dvs. standardavvikelse,
    # istället för varians)
    x1 = stats.norm.rvs(loc=theta, scale=np.sqrt(3))
    x2 = stats.norm.rvs(loc=theta/2, scale=np.sqrt(10))
    
    # Till sist, om vi har x1 och x1 kan vi sedan generera y.
    y = stats.norm.rvs(loc=x1*x2, scale=1)
    
    # Lägg till alla utfall i de motsvarande vektorerna.
    thetas.append(theta)
    x1s.append(x1)
    x2s.append(x2)
    ys.append(y)
```

```python
# Kan nu skatta väntevärdet med hjälp av medelvärdet hos våra utfall.
observed = np.mean(ys)
```

```python
# Väntevärdet som vi redan beräknat analytiskt
expected = 15/16
```
